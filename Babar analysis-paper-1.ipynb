```python
import os
import random
import warnings
import numpy as np
import pandas as pd

from scipy.stats import pearsonr

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import mean_squared_error, r2_score
from sklearn.impute import KNNImputer

import torch
import torch.nn as nn
from torch.utils.data import TensorDataset, DataLoader

warnings.filterwarnings("ignore")

# ============================================================
# PATHS
# ============================================================

BASE = r"C:\Users\sufi\Downloads\Babar"

LABEL_PATH = os.path.join(BASE, "Pan-cancer_label_num.csv")
MRNA_PATH  = os.path.join(BASE, "Pan-cancer_mRNA.csv")
CNV_PATH   = os.path.join(BASE, "Pan-cancer_CNV.csv")

OUTPUT_DIR = os.path.join(BASE, "mandatory_analysis")
os.makedirs(OUTPUT_DIR, exist_ok=True)

# ============================================================
# REPRODUCIBILITY
# ============================================================

SEED = 42

random.seed(SEED)
np.random.seed(SEED)
torch.manual_seed(SEED)

if torch.cuda.is_available():
    torch.cuda.manual_seed_all(SEED)

device = torch.device("cuda" if torch.cuda.is_available() else "cpu")

print("Device:", device)

# ============================================================
# LOAD DATA
# ============================================================

labels_df = pd.read_csv(LABEL_PATH)
mrna_raw  = pd.read_csv(MRNA_PATH)
cnv_raw   = pd.read_csv(CNV_PATH)

print("Labels:", labels_df.shape)
print("mRNA:", mrna_raw.shape)
print("CNV:", cnv_raw.shape)

# ============================================================
# SAMPLE IDENTIFIERS
# ============================================================

mrna_samples = mrna_raw.columns[1:].tolist()
cnv_samples  = cnv_raw.columns[1:].tolist()

assert mrna_samples == cnv_samples
assert len(labels_df) == len(mrna_samples)

sample_ids = np.array(mrna_samples)
labels = labels_df["Label"].values

# ============================================================
# FEATURE IDENTIFIERS
# ============================================================

mrna_genes = mrna_raw.iloc[:, 0].astype(str).values

cnv_gene_raw = cnv_raw.iloc[:, 0].astype(str)

# for enrichment:
# SNORA73|ENSG... -> SNORA73
cnv_gene_clean = (
    cnv_gene_raw
    .str.split("|", regex=False)
    .str[0]
    .str.strip()
    .values
)

# ============================================================
# TRANSPOSE
# Current files = genes x samples
# Models require = samples x genes
# ============================================================

X = mrna_raw.iloc[:, 1:].T.astype(np.float32).values
Y = cnv_raw.iloc[:, 1:].T.astype(np.float32).values

print("X mRNA:", X.shape)
print("Y CNV:", Y.shape)

# ============================================================
# TRAIN / VALIDATION / TEST
# 60% / 15% / 25%
# ============================================================

indices = np.arange(len(labels))

train_idx, temp_idx = train_test_split(
    indices,
    test_size=0.40,
    random_state=SEED,
    stratify=labels
)

val_idx, test_idx = train_test_split(
    temp_idx,
    test_size=0.625,       # 25 / 40 = .625
    random_state=SEED,
    stratify=labels[temp_idx]
)

print("Train:", len(train_idx))
print("Validation:", len(val_idx))
print("Test:", len(test_idx))

# ============================================================
# STANDARDIZATION
# FIT ONLY ON TRAINING DATA
# ============================================================

x_scaler = StandardScaler()
y_scaler = StandardScaler()

X_train = x_scaler.fit_transform(X[train_idx]).astype(np.float32)
X_val   = x_scaler.transform(X[val_idx]).astype(np.float32)
X_test  = x_scaler.transform(X[test_idx]).astype(np.float32)

Y_train = y_scaler.fit_transform(Y[train_idx]).astype(np.float32)
Y_val   = y_scaler.transform(Y[val_idx]).astype(np.float32)
Y_test  = y_scaler.transform(Y[test_idx]).astype(np.float32)

test_sample_ids = sample_ids[test_idx]

print("Prepared.")
```

```python
# ============================================================
# FULL CROSS-MODAL MODEL
# 3217 -> 512 -> 128 -> 512 -> 3105
# ============================================================

class CrossModalNet(nn.Module):
    def __init__(
        self,
        input_dim,
        output_dim,
        hidden1=512,
        bottleneck=128,
        hidden2=512,
        nonlinear=True
    ):
        super().__init__()

        activation = nn.ReLU if nonlinear else nn.Identity

        self.model = nn.Sequential(
            nn.Linear(input_dim, hidden1),
            activation(),
            nn.Linear(hidden1, bottleneck),
            activation(),
            nn.Linear(bottleneck, hidden2),
            activation(),
            nn.Linear(hidden2, output_dim)
        )

    def forward(self, x):
        return self.model(x)


def train_model(
    model,
    X_train,
    Y_train,
    X_val,
    Y_val,
    epochs=10,
    batch_size=128,
    learning_rate=0.001,
    patience=3
):

    model = model.to(device)

    train_dataset = TensorDataset(
        torch.tensor(X_train),
        torch.tensor(Y_train)
    )

    train_loader = DataLoader(
        train_dataset,
        batch_size=batch_size,
        shuffle=True
    )

    X_val_t = torch.tensor(X_val).to(device)
    Y_val_t = torch.tensor(Y_val).to(device)

    optimizer = torch.optim.Adam(
        model.parameters(),
        lr=learning_rate
    )

    criterion = nn.MSELoss()

    best_loss = np.inf
    best_state = None
    patience_counter = 0

    for epoch in range(epochs):

        model.train()

        epoch_losses = []

        for xb, yb in train_loader:

            xb = xb.to(device)
            yb = yb.to(device)

            optimizer.zero_grad()

            pred = model(xb)

            loss = criterion(pred, yb)

            loss.backward()

            optimizer.step()

            epoch_losses.append(loss.item())

        model.eval()

        with torch.no_grad():
            val_pred = model(X_val_t)
            val_loss = criterion(val_pred, Y_val_t).item()

        print(
            f"Epoch {epoch+1:02d} | "
            f"Train={np.mean(epoch_losses):.5f} | "
            f"Val={val_loss:.5f}"
        )

        if val_loss < best_loss:

            best_loss = val_loss

            best_state = {
                key: value.detach().cpu().clone()
                for key, value in model.state_dict().items()
            }

            patience_counter = 0

        else:

            patience_counter += 1

            if patience_counter >= patience:
                print("Early stopping.")
                break

    model.load_state_dict(best_state)

    return model
```

```python
# ============================================================
# REQUIRED BEFORE GENE-LEVEL ANALYSIS
# Train neural model and create neural_pred
# ============================================================

full_model = CrossModalNet(
    input_dim=X_train.shape[1],
    output_dim=Y_train.shape[1],
    hidden1=512,
    bottleneck=128,
    hidden2=512,
    nonlinear=True
)

full_model = train_model(
    full_model,
    X_train,
    Y_train,
    X_val,
    Y_val,
    epochs=10,
    batch_size=128,
    learning_rate=0.001,
    patience=3
)

full_model.eval()

with torch.no_grad():
    neural_pred = full_model(
        torch.tensor(
            X_test,
            dtype=torch.float32
        ).to(device)
    ).cpu().numpy()

print("neural_pred shape:", neural_pred.shape)
print("NaNs:", np.isnan(neural_pred).sum())
print("Infs:", np.isinf(neural_pred).sum())
```

```python
# ============================================================
# FUNCTIONAL ENRICHMENT
# STEP 1: Rank genes by Neural reconstruction quality
# ============================================================

import numpy as np
import pandas as pd
from scipy.stats import pearsonr
from sklearn.metrics import mean_squared_error

neural_gene_results = []

for j in range(Y_test.shape[1]):

    gene = cnv_gene_clean[j]
    raw_gene = cnv_gene_raw.iloc[j]

    gene_mask = mask[:, j]
    n_masked = int(gene_mask.sum())

    if n_masked < 10:
        continue

    y_true = Y_test[gene_mask, j]
    y_pred = neural_pred[gene_mask, j]

    mse = mean_squared_error(
        y_true,
        y_pred
    )

    if (
        np.std(y_true) > 0 and
        np.std(y_pred) > 0
    ):
        r = pearsonr(
            y_true,
            y_pred
        )[0]
    else:
        r = np.nan

    neural_gene_results.append({
        "gene": gene,
        "raw_identifier": raw_gene,
        "n_masked": n_masked,
        "neural_r": r,
        "neural_mse": mse
    })

neural_gene_results = pd.DataFrame(
    neural_gene_results
)

neural_gene_results = (
    neural_gene_results
    .replace([np.inf, -np.inf], np.nan)
    .dropna(subset=["neural_r", "neural_mse"])
)

print(
    "Genes evaluated:",
    len(neural_gene_results)
)

print(
    "Median Neural Pearson r:",
    neural_gene_results["neural_r"].median()
)

print(
    "Median Neural MSE:",
    neural_gene_results["neural_mse"].median()
)

print(
    "\nGenes with r >= 0.50:",
    (neural_gene_results["neural_r"] >= 0.50).sum()
)

print(
    "Genes with r >= 0.60:",
    (neural_gene_results["neural_r"] >= 0.60).sum()
)

print(
    "Genes with r >= 0.70:",
    (neural_gene_results["neural_r"] >= 0.70).sum()
)

print("\nTop 20 genes by Neural Pearson r:")

display(
    neural_gene_results
    .sort_values(
        "neural_r",
        ascending=False
    )
    .head(20)
)
```

```python
# ============================================================
# FUNCTIONAL ENRICHMENT
# STEP 2: Create robust reconstruction-quality ranking
# ============================================================

enrichment_rank = neural_gene_results.copy()

# Higher Pearson r = better
enrichment_rank["rank_r"] = (
    enrichment_rank["neural_r"]
    .rank(method="average", pct=True)
)

# Lower MSE = better
enrichment_rank["rank_mse"] = (
    (-enrichment_rank["neural_mse"])
    .rank(method="average", pct=True)
)

# Equal-weight rank-based quality score
enrichment_rank["quality_score"] = (
    enrichment_rank["rank_r"] +
    enrichment_rank["rank_mse"]
) / 2

# Remove duplicated cleaned gene symbols
# Keep the best-performing occurrence if duplicates exist
enrichment_rank = (
    enrichment_rank
    .sort_values("quality_score", ascending=False)
    .drop_duplicates(subset="gene", keep="first")
    .reset_index(drop=True)
)

print("Unique genes available for enrichment:",
      len(enrichment_rank))

print("\nQuality score summary:")
print(enrichment_rank["quality_score"].describe())

print("\nTop 30 reconstruction-quality genes:")

display(
    enrichment_rank[
        [
            "gene",
            "neural_r",
            "neural_mse",
            "rank_r",
            "rank_mse",
            "quality_score"
        ]
    ].head(30)
)

# Save ranking
rank_path = os.path.join(
    OUTPUT_DIR,
    "functional_enrichment_gene_ranking.csv"
)

enrichment_rank.to_csv(
    rank_path,
    index=False
)

print("\nSaved:", rank_path)
```

```python
# ============================================================
# SURVIVAL / PROGNOSTIC ANALYSIS
# STEP 6: Train-defined risk groups on held-out TEST cohort
# ============================================================

import numpy as np
import pandas as pd
from lifelines import CoxPHFitter


# ------------------------------------------------------------
# Training risk scores
# ------------------------------------------------------------

orig_train_risk = orig_cox.predict_partial_hazard(
    orig_train_cox
).values.ravel()

neural_train_risk = neural_cox.predict_partial_hazard(
    neural_train_cox
).values.ravel()


# ------------------------------------------------------------
# Cancer-specific TRAINING median thresholds
# ------------------------------------------------------------

orig_thresholds = {}
neural_thresholds = {}

for cancer in np.unique(cancer_train):

    idx = np.where(cancer_train == cancer)[0]

    orig_thresholds[cancer] = np.median(
        orig_train_risk[idx]
    )

    neural_thresholds[cancer] = np.median(
        neural_train_risk[idx]
    )


# ------------------------------------------------------------
# Apply frozen thresholds to TEST patients
# ------------------------------------------------------------

orig_test_group = np.array([
    1 if orig_test_risk[i] >= orig_thresholds[cancer_test[i]]
    else 0
    for i in range(len(cancer_test))
])

neural_test_group = np.array([
    1 if neural_test_risk[i] >= neural_thresholds[cancer_test[i]]
    else 0
    for i in range(len(cancer_test))
])


print("ORIGINAL CNV")
print("Low risk:", (orig_test_group == 0).sum())
print("High risk:", (orig_test_group == 1).sum())
print(
    "Deaths low risk:",
    event_test[orig_test_group == 0].sum()
)
print(
    "Deaths high risk:",
    event_test[orig_test_group == 1].sum()
)


print("\nNEURAL RECONSTRUCTED CNV")
print("Low risk:", (neural_test_group == 0).sum())
print("High risk:", (neural_test_group == 1).sum())
print(
    "Deaths low risk:",
    event_test[neural_test_group == 0].sum()
)
print(
    "Deaths high risk:",
    event_test[neural_test_group == 1].sum()
)


# ============================================================
# Held-out TEST Cox validation
# One binary risk-group predictor
# Stratified by cancer type
# ============================================================

def test_group_cox(group):

    df = pd.DataFrame({
        "OS_time": time_test,
        "OS_event": event_test,
        "risk_group": group,
        "cancer_type": cancer_test.astype(str)
    })

    cph = CoxPHFitter()

    cph.fit(
        df,
        duration_col="OS_time",
        event_col="OS_event",
        strata=["cancer_type"]
    )

    return cph


orig_group_cox = test_group_cox(
    orig_test_group
)

neural_group_cox = test_group_cox(
    neural_test_group
)


# ------------------------------------------------------------
# Extract hazard ratios
# ------------------------------------------------------------

orig_hr = np.exp(
    orig_group_cox.params_["risk_group"]
)

orig_ci_low = np.exp(
    orig_group_cox.confidence_intervals_.loc[
        "risk_group", "95% lower-bound"
    ]
)

orig_ci_high = np.exp(
    orig_group_cox.confidence_intervals_.loc[
        "risk_group", "95% upper-bound"
    ]
)

orig_p = orig_group_cox.summary.loc[
    "risk_group", "p"
]


neural_hr = np.exp(
    neural_group_cox.params_["risk_group"]
)

neural_ci_low = np.exp(
    neural_group_cox.confidence_intervals_.loc[
        "risk_group", "95% lower-bound"
    ]
)

neural_ci_high = np.exp(
    neural_group_cox.confidence_intervals_.loc[
        "risk_group", "95% upper-bound"
    ]
)

neural_p = neural_group_cox.summary.loc[
    "risk_group", "p"
]


print("\n===================================")
print("HELD-OUT TEST SURVIVAL VALIDATION")
print("===================================")

print("\nOriginal CNV")
print("HR:", orig_hr)
print("95% CI:", orig_ci_low, "-", orig_ci_high)
print("p-value:", orig_p)

print("\nNeural reconstructed CNV")
print("HR:", neural_hr)
print("95% CI:", neural_ci_low, "-", neural_ci_high)
print("p-value:", neural_p)
```

```python
# ============================================================
# ABLATION STUDY
# Full cross-modal model vs architectural variants
# 5 independent training seeds
# ============================================================

import os
import random
import numpy as np
import pandas as pd
import torch
import torch.nn as nn

from torch.utils.data import TensorDataset, DataLoader
from sklearn.metrics import mean_squared_error, r2_score
from scipy.stats import pearsonr


# ============================================================
# 1. REPRODUCIBILITY
# ============================================================

ABLATION_SEEDS = [42, 123, 456, 789, 1011]

def set_seed(seed):

    random.seed(seed)
    np.random.seed(seed)
    torch.manual_seed(seed)

    if torch.cuda.is_available():
        torch.cuda.manual_seed_all(seed)

    # More deterministic GPU behaviour
    torch.backends.cudnn.deterministic = True
    torch.backends.cudnn.benchmark = False


# ============================================================
# 2. MODEL DEFINITIONS
# ============================================================

class FullModel(nn.Module):
    """
    3217 -> 512 -> 128 -> 512 -> 3105
    ReLU nonlinearities
    """
    def __init__(self, input_dim, output_dim):
        super().__init__()

        self.net = nn.Sequential(
            nn.Linear(input_dim, 512),
            nn.ReLU(),

            nn.Linear(512, 128),
            nn.ReLU(),

            nn.Linear(128, 512),
            nn.ReLU(),

            nn.Linear(512, output_dim)
        )

    def forward(self, x):
        return self.net(x)


class NoNonlinearityModel(nn.Module):
    """
    Same dimensions as Full model,
    but removes all ReLU nonlinearities.
    """
    def __init__(self, input_dim, output_dim):
        super().__init__()

        self.net = nn.Sequential(
            nn.Linear(input_dim, 512),
            nn.Linear(512, 128),
            nn.Linear(128, 512),
            nn.Linear(512, output_dim)
        )

    def forward(self, x):
        return self.net(x)


class ShallowModel(nn.Module):
    """
    3217 -> 128 -> 3105
    Tests importance of depth.
    """
    def __init__(self, input_dim, output_dim):
        super().__init__()

        self.net = nn.Sequential(
            nn.Linear(input_dim, 128),
            nn.ReLU(),
            nn.Linear(128, output_dim)
        )

    def forward(self, x):
        return self.net(x)


class Bottleneck64Model(nn.Module):
    """
    Narrower latent bottleneck.
    3217 -> 512 -> 64 -> 512 -> 3105
    """
    def __init__(self, input_dim, output_dim):
        super().__init__()

        self.net = nn.Sequential(
            nn.Linear(input_dim, 512),
            nn.ReLU(),

            nn.Linear(512, 64),
            nn.ReLU(),

            nn.Linear(64, 512),
            nn.ReLU(),

            nn.Linear(512, output_dim)
        )

    def forward(self, x):
        return self.net(x)


class Bottleneck256Model(nn.Module):
    """
    Wider latent bottleneck.
    3217 -> 512 -> 256 -> 512 -> 3105
    """
    def __init__(self, input_dim, output_dim):
        super().__init__()

        self.net = nn.Sequential(
            nn.Linear(input_dim, 512),
            nn.ReLU(),

            nn.Linear(512, 256),
            nn.ReLU(),

            nn.Linear(256, 512),
            nn.ReLU(),

            nn.Linear(512, output_dim)
        )

    def forward(self, x):
        return self.net(x)


class NoBottleneckModel(nn.Module):
    """
    Removes compact latent bottleneck.
    3217 -> 512 -> 512 -> 3105
    """
    def __init__(self, input_dim, output_dim):
        super().__init__()

        self.net = nn.Sequential(
            nn.Linear(input_dim, 512),
            nn.ReLU(),

            nn.Linear(512, 512),
            nn.ReLU(),

            nn.Linear(512, output_dim)
        )

    def forward(self, x):
        return self.net(x)


# ============================================================
# 3. TRAINING FUNCTION
# ============================================================

def train_ablation_model(
    model,
    X_train,
    Y_train,
    X_val,
    Y_val,
    seed,
    epochs=10,
    batch_size=128,
    learning_rate=0.001,
    patience=3
):

    set_seed(seed)

    model = model.to(device)

    train_dataset = TensorDataset(
        torch.tensor(X_train, dtype=torch.float32),
        torch.tensor(Y_train, dtype=torch.float32)
    )

    # Explicit seeded generator for DataLoader shuffle
    generator = torch.Generator()
    generator.manual_seed(seed)

    train_loader = DataLoader(
        train_dataset,
        batch_size=batch_size,
        shuffle=True,
        generator=generator
    )

    X_val_t = torch.tensor(
        X_val,
        dtype=torch.float32,
        device=device
    )

    Y_val_t = torch.tensor(
        Y_val,
        dtype=torch.float32,
        device=device
    )

    optimizer = torch.optim.Adam(
        model.parameters(),
        lr=learning_rate
    )

    criterion = nn.MSELoss()

    best_loss = np.inf
    best_state = None
    best_epoch = None

    patience_counter = 0

    for epoch in range(epochs):

        model.train()

        for xb, yb in train_loader:

            xb = xb.to(device)
            yb = yb.to(device)

            optimizer.zero_grad()

            pred = model(xb)

            loss = criterion(pred, yb)

            loss.backward()

            optimizer.step()

        # Validation
        model.eval()

        with torch.no_grad():

            val_pred = model(X_val_t)

            val_loss = criterion(
                val_pred,
                Y_val_t
            ).item()

        if val_loss < best_loss:

            best_loss = val_loss
            best_epoch = epoch + 1

            best_state = {
                k: v.detach().cpu().clone()
                for k, v in model.state_dict().items()
            }

            patience_counter = 0

        else:

            patience_counter += 1

            if patience_counter >= patience:
                break

    model.load_state_dict(best_state)

    return model, best_loss, best_epoch


# ============================================================
# 4. ARCHITECTURES
# ============================================================

model_builders = {

    "Full_512-128-512":
        lambda: FullModel(
            X_train.shape[1],
            Y_train.shape[1]
        ),

    "No_Nonlinearity":
        lambda: NoNonlinearityModel(
            X_train.shape[1],
            Y_train.shape[1]
        ),

    "Shallow_128":
        lambda: ShallowModel(
            X_train.shape[1],
            Y_train.shape[1]
        ),

    "Bottleneck_64":
        lambda: Bottleneck64Model(
            X_train.shape[1],
            Y_train.shape[1]
        ),

    "Bottleneck_256":
        lambda: Bottleneck256Model(
            X_train.shape[1],
            Y_train.shape[1]
        ),

    "No_Bottleneck":
        lambda: NoBottleneckModel(
            X_train.shape[1],
            Y_train.shape[1]
        )
}


# ============================================================
# 5. RUN ABLATION
# ============================================================

ablation_results = []

for model_name, builder in model_builders.items():

    print("\n" + "=" * 80)
    print("MODEL:", model_name)
    print("=" * 80)

    for seed in ABLATION_SEEDS:

        print("Seed:", seed)

        # IMPORTANT:
        # set seed BEFORE model construction
        set_seed(seed)

        model = builder()

        model, best_val_loss, best_epoch = train_ablation_model(
            model,
            X_train,
            Y_train,
            X_val,
            Y_val,
            seed=seed,
            epochs=10,
            batch_size=128,
            learning_rate=0.001,
            patience=3
        )

        # --------------------------------------------
        # Predict held-out test CNV
        # --------------------------------------------

        model.eval()

        with torch.no_grad():

            pred_test = model(
                torch.tensor(
                    X_test,
                    dtype=torch.float32,
                    device=device
                )
            ).cpu().numpy()

        # --------------------------------------------
        # Evaluate SAME intentionally masked entries
        # --------------------------------------------

        y_true = Y_test[mask]
        y_pred = pred_test[mask]

        mse = mean_squared_error(
            y_true,
            y_pred
        )

        r2 = r2_score(
            y_true,
            y_pred
        )

        r = pearsonr(
            y_true,
            y_pred
        )[0]

        # Number of trainable parameters
        n_params = sum(
            p.numel()
            for p in model.parameters()
            if p.requires_grad
        )

        ablation_results.append({

            "model": model_name,
            "seed": seed,

            "best_epoch": best_epoch,
            "best_val_mse": best_val_loss,

            "test_mse": mse,
            "test_r2": r2,
            "test_pearson_r": r,

            "parameters": n_params
        })

        print(
            f"Epoch={best_epoch} | "
            f"MSE={mse:.4f} | "
            f"R2={r2:.4f} | "
            f"r={r:.4f}"
        )


# ============================================================
# 6. RESULTS TABLE
# ============================================================

ablation_results = pd.DataFrame(
    ablation_results
)

print("\nRaw ablation runs:")
display(ablation_results)


# ============================================================
# 7. MEAN ± SD SUMMARY
# ============================================================

ablation_summary = (
    ablation_results
    .groupby("model")
    .agg(
        mse_mean=("test_mse", "mean"),
        mse_sd=("test_mse", "std"),

        r2_mean=("test_r2", "mean"),
        r2_sd=("test_r2", "std"),

        pearson_mean=("test_pearson_r", "mean"),
        pearson_sd=("test_pearson_r", "std"),

        mean_best_epoch=("best_epoch", "mean"),

        parameters=("parameters", "first")
    )
    .reset_index()
)


# Order models logically
order = [
    "Full_512-128-512",
    "No_Nonlinearity",
    "Shallow_128",
    "Bottleneck_64",
    "Bottleneck_256",
    "No_Bottleneck"
]

ablation_summary["model"] = pd.Categorical(
    ablation_summary["model"],
    categories=order,
    ordered=True
)

ablation_summary = (
    ablation_summary
    .sort_values("model")
    .reset_index(drop=True)
)


print("\n======================================")
print("ABLATION SUMMARY")
print("======================================")

display(ablation_summary)


# ============================================================
# 8. DIFFERENCE RELATIVE TO FULL MODEL
# ============================================================

full_mse = ablation_summary.loc[
    ablation_summary["model"]
    == "Full_512-128-512",
    "mse_mean"
].iloc[0]

full_r = ablation_summary.loc[
    ablation_summary["model"]
    == "Full_512-128-512",
    "pearson_mean"
].iloc[0]


ablation_summary["delta_mse_vs_full"] = (
    ablation_summary["mse_mean"]
    - full_mse
)

ablation_summary["delta_r_vs_full"] = (
    ablation_summary["pearson_mean"]
    - full_r
)


print("\nDifferences relative to Full model:")

display(
    ablation_summary[
        [
            "model",
            "mse_mean",
            "mse_sd",
            "delta_mse_vs_full",
            "pearson_mean",
            "pearson_sd",
            "delta_r_vs_full",
            "parameters"
        ]
    ]
)


# ============================================================
# 9. SAVE
# ============================================================

raw_path = os.path.join(
    OUTPUT_DIR,
    "ablation_raw_5seeds.csv"
)

summary_path = os.path.join(
    OUTPUT_DIR,
    "ablation_summary_5seeds.csv"
)

ablation_results.to_csv(
    raw_path,
    index=False
)

ablation_summary.to_csv(
    summary_path,
    index=False
)

print("\nSaved:")
print(raw_path)
print(summary_path)
```
